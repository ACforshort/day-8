```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

# Display the dataset
display(df)

# First 5 rows
print("First 5 Rows:")
display(df.head())

# Dataset information
print("Dataset Information:")
df.info()

# Data selection
print("Selected Columns:")
display(df[["Category", "City", "Product", "Payment Method"]])

# Filtering
print("Orders with Sales Greater Than 500:")
display(df[df["Sales"] > 500])

# Sorting
print("Orders Sorted by Sales:")
display(df.sort_values(by="Sales", ascending=False))

# Total Sales
print("Total Sales:")
print(df["Sales"].sum())

# Average Sales
print("Average Sales:")
print(df["Sales"].mean())

# Highest Sales
print("Highest Sales:")
print(df["Sales"].max())

# Lowest Sales
print("Lowest Sales:")
print(df["Sales"].min())

# Total Quantity Sold
print("Total Quantity Sold:")
print(df["Quantity"].sum())

# Number of Orders
print("Number of Orders:")
print(df.shape[0])

# Sales by Category
print("Sales by Category:")
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales)

# Quantity by Category
print("Quantity Sold by Category:")
category_quantity = df.groupby("Category")["Quantity"].sum().sort_values(ascending=False)
display(category_quantity)

# Average Sales by Category
print("Average Sales by Category:")
category_average = df.groupby("Category")["Sales"].mean().sort_values(ascending=False)
display(category_average)

# Sales by City
print("Sales by City:")
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)
display(city_sales)

# Quantity by City
print("Quantity Sold by City:")
city_quantity = df.groupby("City")["Quantity"].sum().sort_values(ascending=False)
display(city_quantity)

# Sales by Product
print("Sales by Product:")
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
display(product_sales)

# Quantity by Product
print("Quantity Sold by Product:")
product_quantity = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
display(product_quantity)

# Sales by Payment Method
print("Sales by Payment Method:")
payment_sales = df.groupby("Payment Method")["Sales"].sum().sort_values(ascending=False)
display(payment_sales)

# Number of Orders by Payment Method
print("Number of Orders by Payment Method:")
payment_orders = df.groupby("Payment Method").size().sort_values(ascending=False)
display(payment_orders)

# Top Performing Products
print("Top 5 Products by Sales:")
display(product_sales.head(5))

# Top Performing Categories
print("Top Categories by Sales:")
display(category_sales.head())

# Highest Sales Orders
print("Top 10 Highest Sales Orders:")
display(df.sort_values(by="Sales", ascending=False).head(10))

# Specific filtering
print("Orders with Quantity Greater Than 5:")
display(df[df["Quantity"] > 5])

# Multiple filtering conditions
print("Orders with Sales Greater Than 500 and Quantity Greater Than 2:")
display(df[(df["Sales"] > 500) & (df["Quantity"] > 2)])

# Aggregation by Category
print("Category-wise Aggregation:")
category_summary = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Highest_Sales=("Sales", "max"),
    Lowest_Sales=("Sales", "min"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Sales", "count")
).sort_values(by="Total_Sales", ascending=False)

display(category_summary)

# Aggregation by City
print("City-wise Aggregation:")
city_summary = df.groupby("City").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Sales", "count")
).sort_values(by="Total_Sales", ascending=False)

display(city_summary)

# Aggregation by Payment Method
print("Payment Method-wise Aggregation:")
payment_summary = df.groupby("Payment Method").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Number_of_Orders=("Sales", "count")
).sort_values(by="Total_Sales", ascending=False)

display(payment_summary)

# Basic Observations
print("Observations:")

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Highest Sale:", df["Sales"].max())
print("Lowest Sale:", df["Sales"].min())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Number of Orders:", df.shape[0])

print("Top Performing Category:", category_sales.index[0])
print("Top Performing Product:", product_sales.index[0])
print("Top Performing City:", city_sales.index[0])
print("Most Used Payment Method:", payment_orders.index[0])
```
